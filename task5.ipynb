TASK-5 Number statistic



```python

print("Number Analysis Program")
# Manually created numerical dataset
numbers = [12, 45, 7, 23, 56, 89, 34, 90, 3, 18]
print("dataset :")
print(numbers)
```

```python
# Create separate lists to store even and odd numbers
even_numbers = []
odd_numbers = []

print("Lists initialized successfully.")
```

```python
# Process each number using a loop
for num in numbers:
    # A number is even if it is divisible by 2
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)
```

```python
# Find the largest and smallest values in the dataset
largest_value = max(numbers)
smallest_value = min(numbers)
```

```python
# Display the analysis results
print("\n" + "=" * 45)
print("           NUMBER ANALYSIS RESULTS")
print("=" * 45)

print(f"Original List  : {numbers}")
print(f"Even Numbers   : {even_numbers}")
print(f"Odd Numbers    : {odd_numbers}")
print(f"Largest Value  : {largest_value}")
print(f"Smallest Value : {smallest_value}")

print("=" * 45)
```
